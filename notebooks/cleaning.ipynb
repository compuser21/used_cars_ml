```python
import pandas as pd
```

```python
df = pd.read_csv('../data/raw/cars.csv')
```

```python
df.head()
df.info()
df.shape
```

```python
print("Shape:", df.shape)
print("\nMissing values:\n", df.isnull().sum())
print("\nDuplicates:", df.duplicated().sum())
print("\nData types:\n", df.dtypes)
```

```python
df.duplicated().sum()
df[df.duplicated()]
```

```python
df = df.drop('reg_year', axis=1)
df.shape
```

```python
df['engine_capacity(CC)'] = df['engine_capacity(CC)'].fillna(df['engine_capacity(CC)'].mean())
df['engine_capacity(CC)'].isnull().sum()
```

```python
df['spare_key']
```

```python
df['spare_key'] = df['spare_key'].map({'Yes': True, 'No': False})
df['spare_key']
```

```python
print("Shape:", df.shape)
print("\nMissing values:\n", df.isnull().sum())
print("\nDuplicates:", df.duplicated().sum())
print("\nData types:\n", df.dtypes)
```

```python
df.to_csv('../data/cleaned/cars.csv')
```
